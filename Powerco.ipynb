```python
price_summary = \
price_df.groupby('id').agg({
    'price_off_peak_var': 'mean',
    'price_peak_var': 'mean',
    'price_mid_peak_var': 'mean',
    'price_off_peak_fix': 'mean',
    'price_peak_fix': 'mean',
    'price_mid_peak_fix': 'mean'
}).reset_index()
```

```python
price_summary.head().T
```

```python
df = client_df.merge(price_summary, on='id', how='left')
df.shape
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)].shape
```

```python
price_cols = [
    'price_off_peak_var',
    'price_peak_var',
    'price_mid_peak_var',
    'price_off_peak_fix',
    'price_peak_fix',
    'price_mid_peak_fix'
]
df[price_cols] = df[price_cols].fillna(df[price_cols].median())
```

```python
for col in ['channel_sales', 'origin_up', 'has_gas']:
  print(f"\n--- {col} ---")
  print(df.groupby(col)
  ['churn'].agg(['mean', 'count']).sort_values('mean', ascending=False))
```
